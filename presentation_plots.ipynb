```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch, Circle

plt.style.use("seaborn-v0_8-muted")

# Pontos do tripleto
x = np.array([0.0+1.0, 0.0+1.0])
y = np.array([1.2+1.0, 0.4+1.0])
z = np.array([2.6+1.0, 1.8+1.0])

fig, ax = plt.subplots(figsize=(6.5, 6.5))

# ---- Estética dos pontos ----
ax.scatter(*x, s=180, color="#1f77b4", edgecolor="black", linewidth=1.2, zorder=3)
ax.scatter(*y, s=180, color="#2ca02c", edgecolor="black", linewidth=1.2, zorder=3)
ax.scatter(*z, s=180, color="#d62728", edgecolor="black", linewidth=1.2, zorder=3)

# ---- Setas representando distâncias ----
arrow_style = dict(arrowstyle="-|>", lw=2.0, mutation_scale=20)

# IMPORTANTE: usar tuplas para os pontos
ax.add_patch(FancyArrowPatch(
    posA=(x[0], x[1]), posB=(y[0], y[1]),
    color="#2ca02c", **arrow_style
))

ax.add_patch(FancyArrowPatch(
    posB=(x[0], x[1]), posA=(y[0], y[1]),
    color="#2ca02c", **arrow_style
))

ax.add_patch(FancyArrowPatch(
    posA=(x[0], x[1]), posB=(z[0], z[1]),
    color="#d62728", **arrow_style
))
ax.add_patch(FancyArrowPatch(
    posB=(x[0], x[1]), posA=(z[0], z[1]),
    color="#d62728", **arrow_style
))

d_xy = np.linalg.norm(x - y)
# ---- Círculo de raio d(x,y) ----
circle = Circle(
    xy=(x[0], x[1]),
    radius=d_xy,
    edgecolor="#1f77b4",
    facecolor="none",
    linewidth=2,
    linestyle="--",
    alpha=0.8,
)
ax.add_patch(circle)



# ---- Rótulos dos pontos ----
ax.text(x[0]-0.1, x[1]-0.18,
        r"$x$ (âncora)",
        fontsize=16, weight="bold", color="#1f77b4")
ax.text(y[0]+0.05, y[1]+0.05, r"$y$ (positivo)", fontsize=15, weight="bold", color="#2ca02c")
ax.text(z[0]-0.95, z[1]+0.05, r"$z$ (negativo)", fontsize=15, weight="bold", color="#d62728")

# ---- Rótulos das distâncias ----
ax.text((x[0]+y[0])/2 - 0.05,
        (x[1]+y[1])/2 - 0.10,
        r"$d(x, y)$", fontsize=14, color="#2ca02c")

ax.text((x[0]+z[0])/2 + 0.09,
        (x[1]+z[1])/2 + 0.00,
        r"$d(x, z)$", fontsize=14, color="#d62728")

# ---- Estética tipo paper ----
ax.set_title("Representação do Triplet",
             fontsize=18, weight="bold", pad=20)

ax.set_xlim(-0.6+1.0, 3.2+1.0)
ax.set_ylim(-0.6+1.0, 2.4+1.0)
ax.set_aspect("equal", "box")
ax.grid(False)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch, Circle

# --- Ativa renderização LaTeX (opcional, mas melhora fonte) ---
plt.rcParams["text.usetex"] = False        # True se tiver LaTeX instalado
plt.rcParams["mathtext.fontset"] = "cm"    # usa Computer Modern (igual ao LaTeX)
plt.rcParams["font.family"] = "serif"

# Pontos do tripleto hard
x = np.array([0.0+1.0, 0.0+1.0])        # âncora
y = np.array([2.2+1.0, 0.3+1.0])        # positivo hard (longe)
z = np.array([0.8+1.0, 0.4+1.0])        # negativo hard (perto)

# Distâncias
d_xy = np.linalg.norm(x - y)
d_xz = np.linalg.norm(x - z)

fig, ax = plt.subplots(figsize=(7, 7))

# ---- Pontos ----
ax.scatter(*x, s=180, color="#1f77b4", edgecolor="black", linewidth=1.2, zorder=3)
ax.scatter(*y, s=180, color="#2ca02c", edgecolor="black", linewidth=1.2, zorder=3)
ax.scatter(*z, s=180, color="#d62728", edgecolor="black", linewidth=1.2, zorder=3)

# ---- Círculo mostrando "região de similaridade" ----
circle = Circle(
    xy=(x[0], x[1]),
    radius=d_xy,
    edgecolor="#1f77b4",
    facecolor="none",
    linestyle="--",
    linewidth=2
)
ax.add_patch(circle)

# ---- Setas ----
arrow_style = dict(arrowstyle="-|>", lw=2.2, mutation_scale=20)

ax.add_patch(FancyArrowPatch(posA=(x[0], x[1]), posB=(y[0], y[1]),
                             color="#2ca02c", **arrow_style))
ax.add_patch(FancyArrowPatch(posA=(x[0], x[1]), posB=(z[0], z[1]),
                             color="#d62728", **arrow_style))

# ---- Rótulos corrigidos ----
ax.text(x[0]-0.12, x[1]-0.17,
        r"$x$ (âncora)",
        fontsize=16, weight="bold", color="#1f77b4")

ax.text(y[0]+0.05, y[1]+0.05,
        r"$y$ (positivo)",
        fontsize=16, weight="bold", color="#2ca02c")

ax.text(z[0]+0.05, z[1]+0.05,
        r"$z$ (negativo)",
        fontsize=16, weight="bold", color="#d62728")

# ---- Rótulos das distâncias ----
ax.text(
    (x[0]+y[0])/2,
    (x[1]+y[1])/2 - 0.09,
    r"$d(x, y)$",
    fontsize=15, color="#2ca02c"
)

ax.text(
    (x[0]+z[0])/2 - 0.3,
    (x[1]+z[1])/2 + 0.1,
    r"$d(x, z)$",
    fontsize=15, color="#d62728"
)

# ---- Título ----
ax.set_title(
    "Exemplo de um Hard Triplet",
    fontsize=20, weight="bold", pad=20
)

# ---- Limpeza estilo slide ----
ax.set_xlim(-0.6+1.0, 3.0+1.0)
ax.set_ylim(-0.6+1.0, 2.2+1.0)
ax.set_aspect("equal")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-muted")

# Distâncias no eixo 1D
d = np.linspace(0, 4, 400)
w = np.exp(-d)

# --------- Caso 1: Loss baixa (positivo mais perto) ---------
d_y1 = 1.0   # positivo perto
d_z1 = 2.5   # negativo longe

w_y1 = np.exp(-d_y1)
w_z1 = np.exp(-d_z1)

# --------- Caso 2: Loss alta (positivo mais longe) ---------
d_y2 = 2.6   # positivo longe (ruim)
d_z2 = 1.2   # negativo mais perto

w_y2 = np.exp(-d_y2)
w_z2 = np.exp(-d_z2)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# ============================================================
# Plot 1 — Loss baixa
# ============================================================
ax = axes[0]
ax.plot(d, w, linewidth=3, color="#1f77b4")

# Pontos
ax.scatter(d_y1, w_y1, s=160, color="#2ca02c", edgecolor="black", linewidth=1.2)
ax.scatter(d_z1, w_z1, s=160, color="#d62728", edgecolor="black", linewidth=1.2)

# Rótulos
ax.text(d_y1+0.05, w_y1+0.03, r"$y$ (positivo)", fontsize=14, weight="bold", color="#2ca02c")
ax.text(d_z1+0.05, w_z1+0.03, r"$z$ (negativo)", fontsize=14, weight="bold", color="#d62728")

ax.set_title("Caso bom: $d(x,y) < d(x,z)$\nLoss baixa", fontsize=16, weight="bold")
ax.set_xlabel("Distância")
ax.set_ylabel("Peso $e^{-d}$")
ax.set_ylim(0, 1.1)
ax.grid(False)

# ============================================================
# Plot 2 — Loss alta
# ============================================================
ax = axes[1]
ax.plot(d, w, linewidth=3, color="#1f77b4")

# Pontos invertidos (positivo mais longe)
ax.scatter(d_y2, w_y2, s=160, color="#2ca02c", edgecolor="black", linewidth=1.2)
ax.scatter(d_z2, w_z2, s=160, color="#d62728", edgecolor="black", linewidth=1.2)

# Rótulos
ax.text(d_y2+0.05, w_y2+0.03, r"$y$ (positivo)", fontsize=14, weight="bold", color="#2ca02c")
ax.text(d_z2+0.05, w_z2+0.03, r"$z$ (negativo)", fontsize=14, weight="bold", color="#d62728")

ax.set_title("Caso ruim: $d(x,y) > d(x,z)$\nLoss alta", fontsize=16, weight="bold")
ax.set_xlabel("Distância")
ax.set_ylim(0, 1.1)
ax.grid(False)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-muted")

fig, ax = plt.subplots(figsize=(7, 7))

# -----------------------------
# 1. Gerar dataset grande
# -----------------------------
np.random.seed(0)

# Três clusters grandes (dados reais)
cluster1 = np.random.multivariate_normal([1.0, 3.0], [[0.15, 0],[0,0.15]], 80)
cluster2 = np.random.multivariate_normal([3.2, 2.0], [[0.18, 0],[0,0.18]], 80)
cluster3 = np.random.multivariate_normal([2.0, 0.8], [[0.12, 0],[0,0.12]], 80)

data = np.vstack([cluster1, cluster2, cluster3])

# -----------------------------
# 2. Proxies (um por classe)
# -----------------------------
proxy1 = np.array([1.0, 3.0])
proxy2 = np.array([3.2, 2.0])
proxy3 = np.array([2.0, 0.8])

proxies = np.vstack([proxy1, proxy2, proxy3])

# -----------------------------
# 3. Plot do dataset real
# -----------------------------
ax.scatter(
    data[:,0], data[:,1],
    s=60, color="lightgray", edgecolor="none", alpha=0.5, label="Dados (D)"
)

# -----------------------------
# 4. Plot dos proxies (estilo-base)
# -----------------------------
proxy_colors = ["#1f77b4", "#2ca02c", "#d62728"]

for p, c in zip(proxies, proxy_colors):
    ax.scatter(
        p[0], p[1],
        s=260, color=c, edgecolor="black", linewidth=1.3, zorder=3
    )

# -----------------------------
# 5. Linhas conectando proxies aos clusters
# (para reforçar a ideia de “representante”)
# -----------------------------
for cluster, proxy, c in zip([cluster1, cluster2, cluster3],
                             proxies, proxy_colors):

    # linha tracejada até a média verdadeira do cluster
    mean = cluster.mean(axis=0)
    ax.plot(
        [proxy[0], mean[0]], [proxy[1], mean[1]],
        linestyle="--", color=c, linewidth=2, alpha=0.9
    )

# -----------------------------
# 6. Rótulos
# -----------------------------
ax.text(proxy1[0]+0.05, proxy1[1]+0.05, r"Proxy 1", fontsize=14,
        weight="bold", color=proxy_colors[0])
ax.text(proxy2[0]+0.05, proxy2[1]+0.05, r"Proxy 2", fontsize=14,
        weight="bold", color=proxy_colors[1])
ax.text(proxy3[0]+0.05, proxy3[1]+0.05, r"Proxy 3", fontsize=14,
        weight="bold", color=proxy_colors[2])

# -----------------------------
# 7. Título e estética final
# -----------------------------
ax.set_title("Proxies vs Dados Reais",
             fontsize=18, weight="bold", pad=20)

ax.set_xlim(0, 4)
ax.set_ylim(0, 4)
ax.set_aspect("equal", "box")
ax.grid(False)
ax.set_xticks([])
ax.set_yticks([])

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch

plt.style.use("seaborn-v0_8-muted")

fig, ax = plt.subplots(figsize=(7, 7))

np.random.seed(0)

# ============================================================
# 1. GERAR 3 CLUSTERS DE DADOS DO CLIENTE (CINZAS)
# ============================================================
cluster1 = np.random.multivariate_normal([1.0, 3.2], [[0.15, 0],[0,0.15]], 60)
cluster2 = np.random.multivariate_normal([3.0, 2.0], [[0.15, 0],[0,0.15]], 60)
cluster3 = np.random.multivariate_normal([2.0, 1.0], [[0.15, 0],[0,0.15]], 60)

data = np.vstack([cluster1, cluster2, cluster3])

ax.scatter(data[:, 0], data[:, 1],
           s=55, color="lightgray", alpha=0.45, edgecolor="none")

# ============================================================
# 2. PROXIES GLOBAIS (CENTRÓIDES)
# ============================================================
P1 = np.array([1.0, 3.2])
P2 = np.array([3.0, 2.0])
P3 = np.array([2.0, 1.0])

proxies_global = [P1, P2, P3]
proxy_colors = ["#1f77b4", "#2ca02c", "#d62728"]

for p, c in zip(proxies_global, proxy_colors):
    ax.scatter(*p, s=260, color=c, edgecolor="black", linewidth=1.3, zorder=4, alpha=1)
    ax.text(p[0] + 0.05, p[1] + 0.05,
            r"$P^{(k)}$", fontsize=12, color=c)

# ============================================================
# 3. PROXIES ADAPTADOS LOCALMENTE (P + ΔP^(k))
# ============================================================
P1_local = P1 + np.array([0.25, -0.15])
P2_local = P2 + np.array([-0.40, -0.10])
P3_local = P3 + np.array([0.25, 0.00])

proxies_local = [P1_local, P2_local, P3_local]

for p_loc, c in zip(proxies_local, proxy_colors):
    ax.scatter(*p_loc, s=260, color=c, edgecolor="black",
               linewidth=1.3, zorder=3, alpha=0.6)
    ax.text(p_loc[0] + 0.05, p_loc[1] - 0.10,
            "Proxy Global", fontsize=12, alpha=0.6, color=c)

# ============================================================
# 4. VETORES DE DESLOCAMENTO (ΔP^(k))
# ============================================================
arrow_style = dict(arrowstyle="-|>", lw=2.3, mutation_scale=18)

for P_g, P_l, c in zip(proxies_global, proxies_local, proxy_colors):
    ax.add_patch(
        FancyArrowPatch(
            posB=tuple(P_g),
            posA=tuple(P_l),
            color=c,
            **arrow_style
        )
    )

# ------------------------------------------------------------
# Título e estética
# ------------------------------------------------------------
ax.set_title("Adaptação Local dos Proxies",
             fontsize=18, weight="bold", pad=20)

ax.set_xlim(0.3, 3.7)
ax.set_ylim(0.4, 3.8)
ax.set_aspect("equal", "box")
ax.grid(False)
ax.set_xticks([])
ax.set_yticks([])

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch, Circle

plt.style.use("seaborn-v0_8-muted")

fig, ax = plt.subplots(figsize=(7, 7))
np.random.seed(0)

# ============================================================
# 1. Clusters de dados do cliente (cinzas)
# ============================================================
cluster1 = np.random.multivariate_normal([1.0, 3.2], [[0.15, 0],[0,0.15]], 60)
cluster2 = np.random.multivariate_normal([3.0, 2.0], [[0.15, 0],[0,0.15]], 60)
cluster3 = np.random.multivariate_normal([2.0, 1.0], [[0.15, 0],[0,0.15]], 60)

data = np.vstack([cluster1, cluster2, cluster3])
ax.scatter(data[:, 0], data[:, 1], s=55, color="lightgray", alpha=0.45)

# ============================================================
# 2. Proxies globais (P)
# ============================================================
Pg1 = np.array([1.0, 3.2])
Pg2 = np.array([3.0, 2.0])
Pg3 = np.array([2.0, 1.0])
proxies_global = [Pg1, Pg2, Pg3]
proxy_colors = ["#1f77b4", "#2ca02c", "#d62728"]

for p, c in zip(proxies_global, proxy_colors):
    # heatmap → círculo transparente
    for r, alpha in zip([0.45, 0.30, 0.18], [0.15, 0.08, 0.03]):
        circ = Circle(tuple(p), radius=r, facecolor=c, edgecolor=None, alpha=alpha)
        ax.add_patch(circ)

    # proxy global
    ax.scatter(*p, s=260, color=c, alpha=0.55, edgecolor="black", linewidth=1.3)
    ax.text(p[0] + 0.05, p[1] + 0.05, "Proxy Global", fontsize=12, color=c, alpha=0.8)

# ============================================================
# 3. Proxies locais (P + ΔP)
# ============================================================
P1_local = Pg1 + np.array([0.25, -0.15])
P2_local = Pg2 + np.array([-0.40, -0.10])
P3_local = Pg3 + np.array([0.25, 0.00])
proxies_local = [P1_local, P2_local, P3_local]

for p_loc, c in zip(proxies_local, proxy_colors):
    # heatmap local (mais forte)
    for r, alpha in zip([0.55, 0.38, 0.22], [0.22, 0.12, 0.06]):
        circ = Circle(tuple(p_loc), radius=r, facecolor=c, edgecolor=None, alpha=alpha)
        ax.add_patch(circ)

    # proxy local
    ax.scatter(*p_loc, s=260, color=c, alpha=1.0, edgecolor="black", linewidth=1.3)
    ax.text(p_loc[0] + 0.05, p_loc[1] - 0.10, r"$P^{(k)}$", fontsize=12, color=c)

# ============================================================
# 4. Setas de deslocamento ΔP^(k)
# ============================================================
arrow_style = dict(arrowstyle="-|>", lw=2.3, mutation_scale=18)

for Pg, Pl, c in zip(proxies_global, proxies_local, proxy_colors):
    ax.add_patch(
        FancyArrowPatch(
            posA=tuple(Pg),
            posB=tuple(Pl),
            color=c,
            **arrow_style
        )
    )

# ------------------------------------------------------------
# Título e estética
# ------------------------------------------------------------
ax.set_title("Adaptação Local dos Proxies\n(Heatmap de Influência por Proxy)",
             fontsize=18, weight="bold", pad=20)

ax.set_xlim(0.3, 3.7)
ax.set_ylim(0.4, 3.8)
ax.set_aspect("equal", "box")
ax.grid(False)
ax.set_xticks([])
ax.set_yticks([])

plt.tight_layout()
plt.show()
```
