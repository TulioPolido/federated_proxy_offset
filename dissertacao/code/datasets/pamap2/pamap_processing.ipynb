```python
import os
import pandas as pd
import re
```

```python
def load_pamap2_folder(folder_path):
    """
    Loads all PAMAP2 files from a folder into one DataFrame with full 54 columns.
    Includes subject_id but does not drop any sensor data.
    """
    all_data = []

    for filename in os.listdir(folder_path):
        if filename.endswith((".dat", ".txt", ".csv")):
            file_path = os.path.join(folder_path, filename)

            # Extract subject ID from filename
            match = re.search(r'(\d+)', filename)
            if not match:
                print(f"Skipping file (no subject ID found): {filename}")
                continue
            subject_id = int(match.group(1)) - 100

            # Load file (assume whitespace-delimited)
            try:
                df = pd.read_csv(file_path, sep=r'\s+', header=None)
            except Exception:
                df = pd.read_csv(file_path, header=None)

            if df.shape[1] != 54:
                print(f"Skipping file (expected 54 columns, got {df.shape[1]}): {filename}")
                continue

            df['subject_id'] = subject_id
            all_data.append(df)

    full_df = pd.concat(all_data, ignore_index=True)
    return full_df
```

```python
folder_path = './raw'
df = load_pamap2_folder(folder_path)
```

```python
df.columns = ['timestamp', 'activity_id', 
              'acc_x', 'acc_y', 'acc_z',
              'gyro_x', 'gyro_y', 'gyro_z', 
              'mag_x', 'mag_y', 'mag_z',
              'subject_id']
df.head(5)
```

```python
# 1. Convert timestamp to long (int64)
df['timestamp'] = df['timestamp'].astype('float64')
df.head(5)
```

```python
# 2. Activity ID to name mapping
activity_map = {
    1: "lying",
    2: "sitting",
    3: "standing",
    4: "walking",
    5: "running",
    6: "cycling",
    7: "Nordic walking",
    9: "watching TV",
    10: "computer work",
    11: "car driving",
    12: "ascending stairs",
    13: "descending stairs",
    16: "vacuum cleaning",
    17: "ironing",
    18: "folding laundry",
    19: "house cleaning",
    20: "playing soccer",
    24: "rope jumping"
    # 0 is excluded on purpose
}

# 3. Filter out activity ID 0
df = df[df['activity_id'] != 0]

# 4. Map activity_id to activity_name
df['activity'] = df['activity_id'].map(activity_map)

# 5. Optional: drop the numeric ID column
df = df.drop(columns=['activity_id'])

df = df.dropna()
# Reset index after all operations
df.reset_index(drop=True, inplace=True)

# Preview result
print(df.head(5))
```

```python
df.to_csv('pamap2_processed.csv', index=False)
```
