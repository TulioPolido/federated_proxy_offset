```python
import pandas as pd
import os
```

```python
# Define the directory containing the dataset files
data_dir = '/home/tulionpl/Downloads/opportunity+activity+recognition/OpportunityUCIDataset/dataset'  # Update with your directory path
```

```python
# List of files to concatenate along with their respective user IDs
files = [
    (f'S{n}-ADL{i}.dat', i) for n in range(1, 5) for i in range(1, 6)
] + [
    (f'S{n}-Drill.dat', n) for n in range(1, 5)
]

files
```

```python
# Initialize an empty list to store DataFrames
dfs = []
```

```python
# Loop through each file and process it
for file, user_id in files:
    file_path = os.path.join(data_dir, file)
    
    # Load the dataset
    df = pd.read_csv(file_path, header=None, delim_whitespace=True)
    
    # Retain only the specified columns
    columns_to_keep = [0, 89, 90, 91, 92, 93, 94, 95, 96, 97, 243]  # Python indices are 0-based, so subtract 1
    df = df.iloc[:, columns_to_keep]
    
    # Rename columns for clarity
    df.columns = [
        'timestamp',
        'accX', 'accY', 'accZ',
        'gyroX', 'gyroY', 'gyroZ',
        'magneticX', 'magneticY', 'magneticZ',
        'activity'
    ]
    
    # Add the user_id column
    df['user_id'] = user_id
    
    # # Apply the transformations
    # df['accX'] = round(df['accX'] / 9.8 * 1000)
    # df['accY'] = round(df['accY'] / 9.8 * 1000)
    # df['accZ'] = round(df['accZ'] / 9.8 * 1000)
    # df['gyroX'] = round(df['gyroX'] * 1000)
    # df['gyroY'] = round(df['gyroY'] * 1000)
    # df['gyroZ'] = round(df['gyroZ'] * 1000)
    # df['magneticX'] = round(df['magneticX'] * 1000)
    # df['magneticY'] = round(df['magneticY'] * 1000)
    # df['magneticZ'] = round(df['magneticZ'] * 1000)
    
    # Append the processed DataFrame to the list
    dfs.append(df)

# Concatenate all the DataFrames
final_df = pd.concat(dfs, ignore_index=True)
```
