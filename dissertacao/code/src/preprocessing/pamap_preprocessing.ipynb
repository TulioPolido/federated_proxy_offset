```python
import pandas as pd
import os
```

```python
# Define the path to the dataset directory
base_path = "/home/tulionpl/Downloads/pamap2+physical+activity+monitoring/PAMAP2_Dataset/Protocol/"

# Define the subjects and the corresponding IDs
subjects = range(101, 110)  # subjects from 101 to 109
subject_ids = range(1, 10)  # IDs from 1 to 9

# Define the columns for PAMAP2
columns = [
    'timestamp', 'activity_id', 'heart_rate',
    'hand_temp', 'hand_acc_16g_x', 'hand_acc_16g_y', 'hand_acc_16g_z',
    'hand_acc_6g_x', 'hand_acc_6g_y', 'hand_acc_6g_z',
    'hand_gyro_x', 'hand_gyro_y', 'hand_gyro_z',
    'hand_mag_x', 'hand_mag_y', 'hand_mag_z',
]
```

```python
# Initialize an empty DataFrame to hold all the data
all_data = pd.DataFrame()

# Loop through each subject and read the data
for subject, subject_id in zip(subjects, subject_ids):
    file_path = os.path.join(base_path, f"subject{subject}.dat")

    # Read the data file into a DataFrame
    df = pd.read_csv(file_path, delim_whitespace=True, header=None)
    # Select only the relevant columns (hand data and identifiers)
    df = df.iloc[:, :16]  # Assuming the first 18 columns correspond to the hand data and identifiers
    df.columns = columns

    # Drop the 'heart_rate' and 'hand_temp' columns
    df = df.drop(['heart_rate', 'hand_temp'], axis=1)

    # Assign the subject ID
    df['subject_id'] = subject_id

    # Filter the DataFrame for the specified activities
    df_filtered = df[df['activity_id'].isin([2, 3, 4, 5, 12, 13])]

    # Append the filtered data to the all_data DataFrame
    all_data = pd.concat([all_data, df_filtered], ignore_index=True)

```

```python
all_data_cleaned = all_data.dropna()
```
