```python
import pandas as pd
from sklearn.utils import resample

# Define the path to your dataset file
file_path = 'WISDM_ar_v1.1_raw.txt'

# Read the dataset into a pandas DataFrame, skipping bad lines
df = pd.read_csv(file_path, header=None, delimiter=",", on_bad_lines='skip')

# Define column names based on the expected structure of the dataset
df.columns = ["user_id", "activity", "timestamp", "x", "y", "z"]
df['z'] = df['z'].str.rstrip(';').astype(float)
```

```python
# Function to balance the dataset
def balance_dataset(df, class_column):
    # Count the occurrences of each class
    class_counts = df[class_column].value_counts()
    min_count = class_counts.min()

    # Resample each class to have the same number of occurrences as the class with the smallest count
    balanced_df = pd.DataFrame()
    for activity in class_counts.index:
        activity_df = df[df[class_column] == activity]
        resampled_df = resample(activity_df, replace=False, n_samples=min_count, random_state=42)
        balanced_df = pd.concat([balanced_df, resampled_df], ignore_index=True)
    
    return balanced_df

# Balance the dataset
balanced_df = balance_dataset(df, "activity")
```

```python
# Save the balanced dataset to a new file
balanced_df = balanced_df.sort_values(by=['user_id', 'timestamp'])
balanced_df.to_csv('WISDM_ar_v1.1_balanced.csv', index=False)

print(f"Balanced dataset saved to 'WISDM_ar_v1.1_balanced.csv'.")
```
