# Client selection for federated learning with heterogeneous resources in mobile edge
### Nishio, T., & Yonetani, R - 2019 - 1416 citações
### Resumo:
Inspirados pela ineficiência em redes FL causada por nós heterogêneos, isto é, nós com recursos diferentes, os autores propuseram um framework de aprendizado federado que utilizasse as informações de recursos dos nós para que o sistema seja otimizado. Esse framework é chamado FedCS (Federated Learning with Client Selection).

### Pontos principais:
- Base de dados: CIFAR-10, Fashion MNIST
- Algoritmo FedCS:
  1. Operador MEC pede aos clientes informações de seus recursos.
  2. Operador MEC utiliza as informações para determinar os clientes que capazes de cumprir o próximo passo dentro de um limite de tempo.
  3. Os parâmetros do modelo global são distribuídos aos clientes selecionados.
  4. Os clientes atualizam os modelos em paralelo e enviam novos parâmetros ao servidor.
  6. O servidor agrega os modelos recebidos em um novo modelo.
- Os resultados demonstram uma redução grande no tempo para atingir a mesma acurácia quando comparado o modelo padrão e FedCS.

### Pontos positivos:
- Parece ser o primeiro artigo a levar em conta uma seleção de clientes no âmbito de FL.
- Os testes foram feitos para dados IID e Non-IID

# Towards Understanding Biased Client Selection in Federated Learning *
### YJ Cho - 2022 - 110 citações
### Resumo:
O artigo prova que a enviesação da seleção de clientes em direção a clientes com maior Loss aumenta a razão de convergência do modelo. Com isso, os autores propuseram uma nova estratégia de seleção de clientes denominada Power-Of-Choice (PoC) que leva a uma convergência até 3x mais rápida, além de aumentar a performance de testes em 10%.

### Pontos principais:
- Um algoritmo de seleção de clientes que prefere clientes com maior Loss, resulta em uma convergência mais rápida do modelo global.
- Algoritmo Power of Choice:
  1. O servidor central calcula um conjunto A de d clientes, onde cada cliente tem probabilidade de ser escolhido pk (fração de dados daquele cliente em relação ao total).
  2. O servidor envia o modelo global aos clientes que calculam sua Loss e retornam ao servidor.
  3. O servidor seleciona m clientes com maior Loss. Esses participarão do treinamento durante a próxima rodada.
- Bases de dados: FMNIST, EMNINST, CIFAR10

### Pontos positivos:
- Não é um artigo apenas fundado no empirismo. Ele se utiliza de matemática teórica para desenvolvimento do algoritmo.

### Pontos negativos:
- Pouco verboso, o que torna a teoria matemática extremamente complexa de ser entendida.

# PyramidFL: A fine-grained client selection framework for efficient federated learning
### Li et al - 2022 - 74 citações
### Resumo:

O artigo propõe o desenvolvimento de um framework de FL que faça a seleção de clientes para treinamento de forma a priorizar clientes que tenham maior utilidade. Clientes que tiveram seus dados utilizados em uma rodada de treinamento, têm menor chance de serem escolhidos para a próxima

### Pontos principais:
- Bases de dados: OpenImage, GoogleSpeech, StackOverflow, HARBox
- PyramidFL é uma melhoria do modelo Oort que possui duas falhas principais. 1, o algoritmo Oort limita o tamanho do conjunto local de um cliente a ser utilizado no treinamento. 2, o Oort não leva em conta que cada update local tem uma importância diferente para o update global, assim, diversos clientes que têm pouca importância para o modelo global enviam seus dados para o servidor.
- PyramidFL:
  1. Selecionar o melhores clientes baseado em sua utilidade
  2. Cada participante atualiza seus valores de utilidade para ajustar a chance de serem escolhidos na próxima rodada.
  3. Cada participante atualiza a importância de seu modelo
  4. Selecionar os participantes com modelos aptos
  5. Calcular o modelo global 

### Pontos positivos:
- O modelo é bem explicado e analisado com mais de uma base de dados

### Pontos negativos:
- As bases de dados não são naturais de um ambiente de FL, portanto a distribuição pode não ser realista se comparada a um problema real de FL.

### Desafios:
- Desenvolver algoritmos de seleção de clientes que levem em conta cenários onde nem sempre todos os clientes estejam disponíveis e estáveis.

# Client Selection On FL: Principles, Challenges and Opportunities
### Lei Fu - 2023 - 38 citações
### Resumo:
Este paper faz uma leitura do estado da arte da seleção de clientes em aprendizado federado. Os artigos mencionados são organizados de acordo com seus critérios para priorização de clientes na seleção. Foram analisados os seguintes aspectos: Heterogeneidade dos clientes, Priorização dos clientes, Implementação atual, Desafios e Oportunidades futuras.

### Pontos principais:
- Heterogeneidade de clientes:
  - Heterogeneidade de sistema: Representa a diferença da capacidade de computação e capacidade de comunicação entre diferentes dispositivos.
  - Heterogeneidade estatística: Representa a diferença de distribuição dos dados. Há três motivos principais. No primeiro, os dados são massivamente distribuidos, tornando o número de clientes muito maior que a média de pontos de dado por cliente. No segundo, os dados são muito desbalanceados, o que faz com que os clientes tenham quantidades muito diferentes de dados. E o terceiro, dados não-IID, o que faz com que cada cliente não possa representar bem a distribuição geral dos dados.
- Priorização dos clientes:
  - Utilidade estatística:
    - Utilidade baseada em dados: Explora os dados locais de um cliente para medir a usabilidade de seu modelo local para o modelo global. (Essa utilidade pode ser medida pela função de perda) 
    \begin{equation}
      \sum_{k} Loss(k) \tag{1}
    \end{equation}
    
    \begin{equation}
      k: \text{dados de um cliente i} \tag{1}
    \end{equation}

    - Utilidade baseada em modelo: Explora os pesos/gradientes dos modelos para medir sua usabilidade. Modelos com pouca divergência em relação ao modelo global podem ser ignorados.

    \begin{equation}
      Divergência = \frac{1}{|w|} \sum_{j=1}^{|w|} |\frac{w_{ij} - \bar{w}_j}{\bar{w}_j}| \tag{2}
    \end{equation}

    \begin{equation}
      w: \text{pesos do modelo local} \tag{2}
    \end{equation}

    \begin{equation}
      \bar{w}: \text{pesos do modelo global} \tag{2}
    \end{equation}

    \begin{equation}
      w_{ij}: \text{j-ésimo peso do i-ésimo cliente} \tag{2}
    \end{equation}


  - Utilidade de sistema: Diferentes hardwares causam diferentes overheads no sistema. Isso pode levar a um pior performance. Limitações de tempo para processamento podem solucionar o problema.


- Implementação atual:
    - Frameworks mais utilizados: TensorFlow Federated (TFF), FedScale, Leaf, FedML.

### Pontos positivos:
