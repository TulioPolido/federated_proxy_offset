```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


def plot_confusion_matrix_pt(
    cm,
    labels,
    titulo="Matriz de Confusão",
    salvar_em=None,
    figsize=(8, 6),
):
    """
    Plota uma matriz de confusão normalizada.

    Parâmetros
    ----------
    cm : np.ndarray
        Matriz de confusão normalizada (linhas somando 1).
    labels : list
        Nome das classes.
    titulo : str
        Título da figura.
    salvar_em : str ou None
        Caminho para salvar a figura.
    figsize : tuple
        Tamanho da figura.
    """

    cm_percent = cm * 100

    plt.figure(figsize=figsize)

    ax = sns.heatmap(
        cm_percent,
        annot=True,
        fmt=".1f",
        cmap="Blues",
        vmin=0,
        vmax=100,
        linewidths=0.5,
        square=True,
        xticklabels=labels,
        yticklabels=labels,
        cbar_kws={"label": "Porcentagem (%)"},
    )

    ax.set_xlabel("Rótulo Previsto", fontsize=12)
    ax.set_ylabel("Rótulo Verdadeiro", fontsize=12)
    ax.set_title(titulo, fontsize=14, pad=15)

    plt.xticks(rotation=45, ha="right")
    plt.yticks(rotation=0)

    plt.tight_layout()

    if salvar_em is not None:
        plt.savefig(
            salvar_em,
            dpi=300,
            bbox_inches="tight",
        )

    plt.show()
```

```python
cm_best = np.array([
    [81.6,  0.0, 11.0,  0.3,  7.2],
    [ 0.0, 99.1,  0.9,  0.0,  0.0],
    [15.5,  0.3, 77.5,  0.3,  6.4],
    [ 0.0,  0.0,  0.9, 75.7, 23.4],
    [ 0.0,  0.0,  6.8, 32.9, 60.3]
]) / 100

cm_mean = np.array([
    [72.58, 0.03, 18.89, 1.57, 6.93],
    [0.00, 99.39, 0.61, 0.00, 0.00],
    [9.16, 0.81, 77.30, 1.42, 11.40],
    [0.00, 0.00, 0.71, 74.42, 24.87],
    [0.00, 0.00, 6.12, 39.42, 54.47]
]) / 100

labels = [
    'Walking', 
    'Jogging', 
    'Stairs', 
    'Sitting', 
    'Standing'
]

```

```python
plot_confusion_matrix_pt(
    cm_best,
    labels,
    titulo="Matriz de Confusão do Melhor Cliente - FedProff WISDM",
    salvar_em="matriz_confusao.pdf",
)
np.mean(np.diag(cm_best))
```

```python
plot_confusion_matrix_pt(
    cm_mean,
    labels,
    titulo="Matriz de Confusão Média - FedProff WISDM",
    salvar_em="matriz_confusao.pdf",
)
np.mean(np.diag(cm_mean))
```
