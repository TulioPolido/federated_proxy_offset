```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


def plot_confusion_matrix_pt(
    cm,
    labels,
    titulo="Matriz de Confusão",
    salvar_em=None,
    figsize=(8, 6),
):
    """
    Plota uma matriz de confusão normalizada.

    Parâmetros
    ----------
    cm : np.ndarray
        Matriz de confusão normalizada (linhas somando 1).
    labels : list
        Nome das classes.
    titulo : str
        Título da figura.
    salvar_em : str ou None
        Caminho para salvar a figura.
    figsize : tuple
        Tamanho da figura.
    """

    cm_percent = cm * 100

    plt.figure(figsize=figsize)

    ax = sns.heatmap(
        cm_percent,
        annot=True,
        fmt=".1f",
        cmap="Blues",
        vmin=0,
        vmax=100,
        linewidths=0.5,
        square=True,
        xticklabels=labels,
        yticklabels=labels,
        cbar_kws={"label": "Porcentagem (%)"},
    )

    ax.set_xlabel("Rótulo Previsto", fontsize=12)
    ax.set_ylabel("Rótulo Verdadeiro", fontsize=12)
    ax.set_title(titulo, fontsize=14, pad=15)

    plt.xticks(rotation=45, ha="right")
    plt.yticks(rotation=0)

    plt.tight_layout()

    if salvar_em is not None:
        plt.savefig(
            salvar_em,
            dpi=300,
            bbox_inches="tight",
        )

    plt.show()
```

```python
cm_best = np.array([
    [79.3, 0.0, 10.4, 0.9, 9.5],
    [0.0, 99.4, 0.6, 0.0, 0.0],
    [4.6, 0.3, 88.8, 0.6, 5.8],
    [0.0, 0.0, 0.6, 70.2, 29.2],
    [0.0, 0.0, 6.8, 15.4, 77.8]
]) / 100


cm_mean = np.array([
    [81.29, 0.04, 8.87, 0.88, 8.96],
    [0.04, 99.59, 0.37, 0.00, 0.00],
    [11.46, 0.47, 78.89, 1.74, 7.44],
    [0.08, 0.00, 0.99, 72.84, 26.09],
    [1.00, 0.04, 6.11, 23.94, 68.89]
]) / 100


labels = [
    'Walking', 
    'Jogging', 
    'Stairs', 
    'Sitting', 
    'Standing'
]

```

```python
plot_confusion_matrix_pt(
    cm_best,
    labels,
    titulo="Matriz de Confusão do Melhor Cliente - FedMeta + ProxyNCA WISDM",
    salvar_em="matriz_confusao.pdf",
)
np.mean(np.diag(cm_best))
```

```python
plot_confusion_matrix_pt(
    cm_mean,
    labels,
    titulo="Matriz de Confusão Média - FedMeta + ProxyNCA WISDM",
    salvar_em="matriz_confusao.pdf",
)
np.mean(np.diag(cm_mean))
```
