```python
def update_ema(
    prev_ema,
    current_value: float,
    smoothing_weight: float,
) -> float:
    """We use EMA to visually enhance the learning trend for each round.

    Parameters
    ----------
    prev_ema : float
        The list of metrics to aggregate.
    current_value : float
        The list of metrics to aggregate.
    smoothing_weight : float
        The list of metrics to aggregate.


    Returns
    -------
    EMA_Loss or EMA_ACC
        The weighted average metric.
    """
    if prev_ema is None:
        return current_value
    return (1 - smoothing_weight) * current_value + smoothing_weight * prev_ema
```

```python
def compute_ema_series(values, smoothing_weight=0.95):
    ema_values = []
    prev_ema = None

    for v in values:
        prev_ema = update_ema(prev_ema, v, smoothing_weight)
        ema_values.append(prev_ema)

    return ema_values
```

```python
def plot_flower(avg_acc, avg_loss, use_ema=False):
    """
    Plota múltiplos modelos em dois gráficos separados (accuracy e loss)
    """

    if use_ema:
        smoothing_weight = 0.95

        avg_acc = {
            k: pd.Series(
                compute_ema_series(v, smoothing_weight),
                index=v.index
            )
            for k, v in avg_acc.items()
        }

        avg_loss = {
            k: pd.Series(
                compute_ema_series(v, smoothing_weight),
                index=v.index
            )
            for k, v in avg_loss.items()
        }

    # -------------------------
    # ACCURACY FIGURE
    # -------------------------
    plt.figure(figsize=(7, 4.5))  # paper-friendly

    for name, series in avg_acc.items():
        plt.plot(
            series.index.astype(int),
            series.values,
            linewidth=3,
            label=name,
        )

    plt.title("Accuracy")
    plt.ylabel("Average Accuracy")
    plt.xlabel("Round")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()

    # -------------------------
    # LOSS FIGURE
    # -------------------------
    plt.figure(figsize=(7, 4.5))  # same size

    for name, series in avg_loss.items():
        plt.plot(
            series.index.astype(int),
            series.values,
            linewidth=3,
            label=name,
        )

    plt.title("Loss")
    plt.ylabel("Average Loss")
    plt.xlabel("Round")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
files = {
    "./Personalized/FedMeta ProxyNCA V2/Results/PAMAP/measures011A.csv": "FedProff",
    "./Federated/ProxyNCA/Results/PAMAP/measures001A.csv": "FedMeta + ProxyNCA",
    #"./Personalized/FedMeta MobileNetV2/Results/PAMAP/measures024A.csv": "FedMeta + MobileNetV2",
}
```

```python
avg_loss = {}
avg_acc = {}

for path, name in files.items():
    df = pd.read_csv(path)
    df['validation'] = df['validation'].astype(bool)

    df_val = df[df['validation'] == True]

    users = sorted(df['user'].unique())

    all_losses = []
    all_accs = []

    for user in users:
        user_df = df_val[df_val['user'] == user]

        pivot = user_df.pivot(index='epoch', columns='metric', values='values')

        # -------- LOSS --------
        if "metrics.LOSS" in pivot.columns:
            loss = pd.to_numeric(pivot["metrics.LOSS"], errors="coerce")
            loss.index = pd.to_numeric(pivot.index, errors="coerce")
            loss = loss[~loss.index.isna()]

            if not loss.empty:
                loss.name = user
                all_losses.append(loss)

        # -------- ACC --------
        if "metrics.TEST_ACC" in pivot.columns:
            acc = pd.to_numeric(pivot["metrics.TEST_ACC"], errors="coerce")
            acc.index = pd.to_numeric(pivot.index, errors="coerce")
            acc = acc[~acc.index.isna()]

            if not acc.empty:
                acc.name = user
                all_accs.append(acc)

    # -------- média por arquivo --------
    if all_losses:
        df_loss = pd.concat(all_losses, axis=1).sort_index()
        avg_loss[name] = df_loss.mean(axis=1, skipna=True)

    if all_accs:
        df_acc = pd.concat(all_accs, axis=1).sort_index()
        avg_acc[name] = df_acc.mean(axis=1, skipna=True)
```

```python
plt.figure(figsize=(10, 5))

for name, series in avg_loss.items():
    plt.plot(series.index.astype(int), series.values,
             marker='o', label=name)

plt.xlabel('Round')
plt.ylabel('Average Loss')
plt.title('Comparison of Loss between Models')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 5))

for name, series in avg_acc.items():
    plt.plot(series.index.astype(int), series.values,
             marker='o', label=name)

plt.xlabel('Round')
plt.ylabel('Average Accuracy')
plt.ylim(0, 1)
plt.title('Comparison of Accuracy between Models')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plot_flower(avg_acc, avg_loss, use_ema=True)
```
