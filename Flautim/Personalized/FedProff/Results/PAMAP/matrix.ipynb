```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


def plot_confusion_matrix_pt(
    cm,
    labels,
    titulo="Matriz de Confusão",
    salvar_em=None,
    figsize=(8, 6),
):
    """
    Plota uma matriz de confusão normalizada.

    Parâmetros
    ----------
    cm : np.ndarray
        Matriz de confusão normalizada (linhas somando 1).
    labels : list
        Nome das classes.
    titulo : str
        Título da figura.
    salvar_em : str ou None
        Caminho para salvar a figura.
    figsize : tuple
        Tamanho da figura.
    """

    cm_percent = cm * 100

    plt.figure(figsize=figsize)

    ax = sns.heatmap(
        cm_percent,
        annot=True,
        fmt=".1f",
        cmap="Blues",
        vmin=0,
        vmax=100,
        linewidths=0.5,
        square=True,
        xticklabels=labels,
        yticklabels=labels,
        cbar_kws={"label": "Porcentagem (%)"},
    )

    ax.set_xlabel("Rótulo Previsto", fontsize=12)
    ax.set_ylabel("Rótulo Verdadeiro", fontsize=12)
    ax.set_title(titulo, fontsize=14, pad=15)

    plt.xticks(rotation=45, ha="right")
    plt.yticks(rotation=0)

    plt.tight_layout()

    if salvar_em is not None:
        plt.savefig(
            salvar_em,
            dpi=300,
            bbox_inches="tight",
        )

    plt.show()
```

```python
plot_confusion_matrix_pt(
    cm_best,
    labels,
    titulo="Matriz de Confusão do Melhor Cliente - FedProff PAMAP2",
    salvar_em="matriz_confusao.pdf",
)

np.mean(np.diag(cm_best))
```
